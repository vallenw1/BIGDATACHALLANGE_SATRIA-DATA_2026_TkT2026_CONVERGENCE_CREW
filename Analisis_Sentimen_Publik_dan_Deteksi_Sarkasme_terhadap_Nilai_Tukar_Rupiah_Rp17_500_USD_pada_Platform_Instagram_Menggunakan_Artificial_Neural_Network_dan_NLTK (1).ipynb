```python
import re

def clean_text(text):
    # 1. Case Folding: Mengubah ke huruf kecil
    text = text.lower()
    # 2. Menghapus URL/Link
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)
    # 3. Menghapus Mentions (@) dan Hashtags (#)
    text = re.sub(r'@\w+|#\w+', '', text)
    # 4. Menghapus tanda baca dan karakter khusus
    text = re.sub(r'[^\w\s]', '', text)
    # 5. Menghapus angka
    text = re.sub(r'\d+', '', text)
    # 6. Menghapus spasi berlebih
    text = text.strip()
    # 7. Menghapus emoticon dan karakter non-ASCII
    text = re.sub(r'[\U0001F600-\U0001F64F\U0001F300-\U0001F5FF\U0001F680-\U0001F6FF\U0001F1E0-\U0001F1FF]', '', text)
    text = re.sub(r'[^\x00-\x7F]+', '', text) # Remove other non-ASCII characters
    return text

# Menerapkan fungsi pembersihan
df['text_bersih'] = df['text'].apply(clean_text)

# Columns to drop for sentiment analysis, focusing on text content
columns_to_drop = [
    'postId', 'postUrl', 'commentId', 'commentUrl', 'timestamp',
    'likesCount', 'userId', 'username', 'userFullName',
    'ownerProfilePicUrl', 'owner', 'ownerUsername',
    'isVerified', 'hasLikedComment', 'childCommentCount', 'repliesCount', 'replies'
]

df = df.drop(columns=columns_to_drop, errors='ignore')

# Display the DataFrame head to show cleaned text and dropped columns
df
```

# *Training Model*

```python
# Building Model
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GlobalAveragePooling1D, Dense, Dropout, Input
from tensorflow.keras.regularizers import l1, l2

model = Sequential()
# Explicitly define the input shape using an Input layer, as input_length in Embedding is deprecated.
model.add(Input(shape=(max_len,)))
model.add(Embedding(word_voc_length, 150)) # Removed input_length as it's deprecated and now handled by the Input layer
model.add(GlobalAveragePooling1D())
model.add(Dense(20, activation='relu', kernel_regularizer=l2(l2=0.05))) # Kurangi neuron, tingkatkan L2
model.add(Dropout(0.4)) # Tingkatkan dropout
# Ubah lapisan output untuk klasifikasi multi-kelas (3 kelas: negatif, netral, positif)
# Gunakan jumlah unit sesuai dengan jumlah kelas (len(label_mapping)) dan aktivasi 'softmax'
model.add(Dense(len(label_mapping), activation='softmax', kernel_regularizer=l2(l2=0.01)))

# Kompilasi model dengan fungsi loss yang sesuai untuk klasifikasi multi-kelas
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
model.summary()
```

### Addressing Data Imbalance with SMOTE

The current dataset for sarcasm detection is highly imbalanced, with very few 'sarcastic' comments compared to 'non-sarcastic' ones. This makes it difficult for the model to learn to identify the minority class. To address this, we will apply **SMOTE (Synthetic Minority Over-sampling Technique)** to the training data. SMOTE generates synthetic samples for the minority class, helping to balance the class distribution and improve the model's ability to recognize sarcastic comments.

```python
from imblearn.over_sampling import SMOTE
import pandas as pd

print("Original training set shape before SMOTE:", X_train.shape, y_train.shape)
print("Original class distribution in training set:", pd.Series(y_train).value_counts())

# Apply SMOTE to the training data
# Adjusted k_neighbors to be less than or equal to the number of minority samples
# Ensure k_neighbors is valid: min(number of minority samples - 1, default k_neighbors)
minority_class_count = pd.Series(y_train).value_counts().get(1, 0)
k_neighbors_smote = min(3, minority_class_count - 1) if minority_class_count > 1 else 1

# If minority_class_count is 0 or 1, SMOTE cannot be applied or k_neighbors will be invalid
# For such cases, you might want to skip SMOTE or handle it differently.
# For now, let's assume we have at least 2 minority samples for k_neighbors = 1.

if minority_class_count > 1:
    sm = SMOTE(random_state=42, k_neighbors=k_neighbors_smote)
    X_train_res, y_train_res = sm.fit_resample(X_train, y_train)
    print("\nTraining set shape after SMOTE:", X_train_res.shape, y_train_res.shape)
    print("Class distribution in training set after SMOTE:", pd.Series(y_train_res).value_counts())
else:
    print("\nSMOTE skipped: Not enough minority samples (sarcastic) in the training set.")
    X_train_res = X_train
    y_train_res = y_train
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix
from imblearn.over_sampling import SMOTE
import pandas as pd

# 1. Vektorisasi dengan TF-IDF
tfidf = TfidfVectorizer(max_features=1000, ngram_range=(1,2))
X_tfidf = tfidf.fit_transform(df['text_stopword_dihapus']).toarray()
y = df['numeric_sentiment_label'].values

# 2. Split Data dengan Stratify (agar distribusi kelas seimbang di train & test)
X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf, y, test_size=0.2, random_state=42, stratify=y
)

# 3. Terapkan SMOTE pada FITUR TF-IDF (Bukan pada pad_sequences!)
smote = SMOTE(random_state=42, k_neighbors=3)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

print("Distribusi kelas setelah SMOTE:", pd.Series(y_train_res).value_counts())

# 4. Latih Model SVM
model_svm = SVC(kernel='linear', class_weight='balanced', random_state=42)
model_svm.fit(X_train_res, y_train_res)

# 5. Evaluasi Performa
y_pred = model_svm.predict(X_test)
print("\n--- LAPORAN KLASIFIKASI SVM ---")
print(classification_report(y_test, y_pred, target_names=['Negative', 'Neutral', 'Positive']))
```
